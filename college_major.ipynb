Sorting Values & Adding Columns: Majors with the Most Potential vs Lowest Risk

```python
#cleandf['Mid-Career 90th Percentile Salary'] - cleandf['Mid-Career 10th Percentile Salary']
#cleandf['Mid-Career 90th Percentile Salary'].subtract(cleandf['Mid-Career 10th Percentile Salary'])
spread = cleandf['Mid-Career 90th Percentile Salary'] - cleandf['Mid-Career 10th Percentile Salary']
cleandf.insert(1, 'Spread', spread)
low_risk = cleandf.sort_values('Spread')
low_risk[['Undergraduate Major','Spread']].head()
```

```python
potencial = cleandf.sort_values('Mid-Career 90th Percentile Salary', ascending=False)
potencial[['Undergraduate Major','Mid-Career 90th Percentile Salary']].head()
```

```python
potencial = cleandf.sort_values('Mid-Career Median Salary', ascending=False)
potencial[['Undergraduate Major', 'Mid-Career Median Salary']].head()
```
